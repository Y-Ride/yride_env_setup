# Rotation Matrix Debug
Below is code to allow for testing different rotation matrices to compare them.

```python
import numpy as np
is_degrees = True

def Rx(angle):
    if is_degrees:
        angle = np.deg2rad(angle)
    return np.array([[1.0,            0.0,            0.0          ],
                     [0.0,            np.cos(angle), -np.sin(angle)],
                     [0.0,            np.sin(angle),  np.cos(angle)]])

def Ry(angle):
    if is_degrees:
        angle = np.deg2rad(angle)
    return np.array([[np.cos(angle),  0.0,            np.sin(angle)],
                     [0.0,            1.0,            0.0          ],
                     [-np.sin(angle), 0.0,            np.cos(angle)]])

def Rz(angle):
    if is_degrees:
        angle = np.deg2rad(angle)
    return np.array([[np.cos(angle), -np.sin(angle),  0.0],
                     [np.sin(angle),  np.cos(angle),  0.0],
                     [0.0,            0.0,            1.0]])
    
def FinalRot(*matrices):
    '''
    Input -> List of SO(3) rotation matrices
        Ex. FinalRot(Rx(90), Rz(90), Ry(90))
    Output -> Combination of all rotation matrices in order passed in
        Ex. Cont. return Rx @ Rz @ Ry
    Current frame -> postmultiply
        R1 @ R2
    Fixed frame -> premultipy
        R2 @ R1
    '''
    tmp = None
    for mat in matrices:  # reverse multiply order
        if mat.shape != (3,3):
            print(f"Please make sure all matrices are 3x3 rotation matrices")
        if tmp is None:
            tmp = mat
        else:
            tmp = np.dot(tmp, mat)
    return tmp
```

```python
np.set_printoptions(suppress=True)
# Rotations about current frame are post multiplied
# Rotations about fixed frame are pre multiplied
FinalRot(Rx(180), Rz(180))
```
